# outlier handling

```python
feature_df.describe()
```

```python
import numpy as np
features=['noOfRcs', 'noOfTrans', 'riceAfsc', 'riceFsc', 'riceAap',
       'wheat', 'sugar', 'rgdal', 'kerosene', 'totalAmount', 'salt',
       'otherShopTransCnt','rcNfsaAay', 'unitsNfsaAay', 'rcNfsaPhh', 'unitsNfsaPhh',
       'totalRcNfsa', 'totalUnitsNfsa', 'rcStateAay', 'unitsStateAay',
       'rcStatePhh', 'unitsStatePhh', 'rcStateAap', 'unitsStateAap',
       'totalRcState', 'totalUnitsState', 'totalRcs', 'totalUnits',]
for col in features:
    Q1=feature_df[col].quantile(0.25)
    Q3=feature_df[col].quantile(0.75)
    IQR=Q3 - Q1
    LW=Q1-1.5*IQR
    UW=Q3+1.5*IQR
    feature_df[col]=np.where(feature_df[col]<LW,LW,feature_df[col])
    feature_df[col]=np.where(feature_df[col]>UW,UW,feature_df[col])
```

```python
plt.figure(figsize=(35,7))
plt.suptitle('BoxPLot for outlier')
for i in range(0,len(features)):
    plt.subplot(1,28,i+1)
    sns.boxplot(y=feature_df[features[i]],color='purple',)
    plt.tight_layout()
```

# Clustering using PCA on feature columns

```python
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
# Select numeric feature columns
features=['noOfRcs', 'noOfTrans', 'riceAfsc', 'riceFsc', 'riceAap',
       'wheat', 'sugar', 'rgdal', 'kerosene', 'totalAmount', 'salt',
       'otherShopTransCnt','rcNfsaAay', 'unitsNfsaAay', 'rcNfsaPhh', 'unitsNfsaPhh',
       'totalRcNfsa', 'totalUnitsNfsa', 'rcStateAay', 'unitsStateAay',
       'rcStatePhh', 'unitsStatePhh', 'rcStateAap', 'unitsStateAap',
       'totalRcState', 'totalUnitsState', 'totalRcs', 'totalUnits',]
# Extract feature matrix X
x = feature_df[features].copy()
#Scale features
scaler = StandardScaler()
x_scaled = scaler.fit_transform(x)
# Apply PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(x_scaled)

#Create a DataFrame with PCA results
pca_df = pd.DataFrame(data=pca_result, columns=['PC1', 'PC2'])

pca_df.head()
```

```python
print("Explained Variance Ratio:", pca.explained_variance_ratio_)
print("Total Variance Retained:", sum(pca.explained_variance_ratio_))
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=pca_df, x='PC1', y='PC2', alpha=0.5)
plt.title('PCA - First 2 Principal Components')
plt.show()
```

```python
#K-mean Clusturing
```

```python
from sklearn.cluster import KMeans
```

```python
inertia_list=[]
for k in range(2,11):
    kmeans=KMeans(n_clusters=k,random_state=42)
    kmeans.fit(x_scaled)
    inertia_list.append(kmeans.inertia_)
    
```

```python
plt.plot(range(2,11),inertia_list,marker='o')
plt.title('Elbow Method for Optimal k')
plt.show()
```

```python
# Initialize KMeans
k = 4
kmeans = KMeans(n_clusters=k, random_state=42, n_init=10)

# Fit on scaled features and assign cluster labels to main_df
feature_df['Cluster'] = kmeans.fit_predict(x_scaled)

# If you also want cluster labels in pca_df:
pca_df['Cluster'] = feature_df['Cluster']

pca_df.head()
```

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(
    data=pca_df, 
    x='PC1', 
    y='PC2', 
    hue='Cluster', 
    palette='viridis', 
    alpha=0.6
)

# Plot Cluster Centroids on PCA Space
centroids_pca = pca.transform(kmeans.cluster_centers_)
plt.scatter(
    centroids_pca[:, 0], 
    centroids_pca[:, 1], 
    s=200, 
    c='red', 
    marker='X', 
    label='Centroids'
)

plt.title('K-Means Clusters (Visualized via PCA)')
plt.legend()
plt.show()
```

```python
kmeans.inertia_
```

```python
# Group by Cluster to see average values for key features
cluster_summary = feature_df.groupby('Cluster')[features].mean()
print(cluster_summary)
```

# cluster and noise estimation by DBSCAN algorithm

```python
from sklearn.cluster import DBSCAN
newdf=all_feature[['shopNo','latitude','longitude']]
# Instantiate and fit DBSCAN
dbscan = DBSCAN(eps=0.5, min_samples=5)
labels = dbscan.fit_predict(newdf)

# Inspect results
# Label -1 indicates noise / outliers
n_clusters = len(set(labels)) - (1 if -1 in labels else 0)
n_noise = list(labels).count(-1)

print(f"Estimated number of clusters: {n_clusters}")
print(f"Estimated number of noise points: {n_noise}")
```
